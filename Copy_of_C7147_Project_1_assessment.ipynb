## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence:str, reference:str) -> dict:
    """This function checks for sequence validity - correct length,
    valid characters, no empty sequences."""
    sequence = sequence.upper() # Lowercase inputs are valid
    errors = [] # Initialization

    # If the string is empty/length is incorrect, notes error as wrong_length
    if len(sequence) == 0 or len(sequence) != len(reference):
      errors.append("wrong_length")

    #If base is not in ACTG, notes error as invalid_character/appends to errors
    for base in sequence:
      if base not in "ACTG":
        errors.append("invalid_characters")

    # validity is when there are no errors
    valid = len(errors) == 0

    return {"sequence": sequence, "valid": valid, "errors": errors}
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
def find_mutations(sequence: str, reference:str) -> list[dict[int, str, str]]:
    """This function finds mutations - notes position of the mutation, original
    base in reference, and changed base in sequence."""
    mismatch = [] # Initialization

    # for base by base comparison of sequence, reference; numbered
    for position, (base_a, base_b) in enumerate(zip(sequence, reference)):

      # If the bases are not equal, append example for each mismatch and return
      if base_a != base_b:
        example = {"position": position + 1, "reference_base": reference[position], "read_base": sequence[position]}
        mismatch.append(example)

    return mismatch
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads: list, reference:str) -> list[dict]:
    """This function takes in reads and returns a dictionary for each read w/
    read information and validity status, read errors, and mutations."""

    for read in reads:
      read["sequence"] = read["sequence"].upper() # Normalizes sequences

      # Assign validity status and find mutations only for valid sequences
      validate_read(read["sequence"], reference)
      find_mutations(read["sequence"], reference)
      read["status"] = "valid" if validate_read(read["sequence"], reference)["valid"] else "invalid"
      read["errors"] = validate_read(read["sequence"], reference)["errors"]
      if read["status"] == "valid":
        read["mutations"] = find_mutations(read["sequence"], reference)
      else:
        read["mutations"] = []

    return reads
````

### 💻 Enter your code here

````python
def summarise_samples(analysed_reads: list[dict]) -> dict[dict]:
  """This function takes in analysed reads and returns a dictionary per sample
     of read validity, mutation reads and events; and the types and positions
     of those mutations w/ percentages and top positions specified."""

  cyan = {} # Initialize empty dictionary

  # Set purple as a reference for dictionary structure and if sample not present
  # in cyan, append it based on the structure defined in purple
  purple = {"group" : "", "valid_reads" : 0, "invalid_reads" : 0,
            "mismatch_reads" : 0, "mismatch_events" : 0,
            "mismatch_percentage" : 0, "substitution_counts" : {},
            "position_counts" : {}, "top_positions" : []}
  for sample in analysed_reads:
    if sample['sample'] not in cyan:
      cyan[sample['sample']] = {"group" : "", "valid_reads" : 0,
                                "invalid_reads" : 0, "mismatch_reads" : 0,
                                "mismatch_events" : 0, "mismatch_percentage" : 0,
                                "substitution_counts" : {}, "position_counts" : {},
                                "top_positions" : []}

  # For loop has to be repeated for each different task so cyan doesn't copy
  # values of the second sample in purple for every other sample it encounters
  for sample in analysed_reads:
    purple = cyan[sample["sample"]]

  # Define groups and if sample is valid, count 1 for each, else invalid
    purple["group"] = sample["group"]
    if sample["status"] == "valid":
      purple["valid_reads"] += 1
    else:
      purple["invalid_reads"] +=1

  # If mutations is not empty, add one to mismatch reads, and add the length
  # of the mutation for mismatch events and calculate mismatch percentage
  # from valid_reads only- analyse_reads only records mutations for valid reads
  for sample in analysed_reads:
    purple = cyan[sample["sample"]]
    if sample["mutations"] != []:
      purple["mismatch_reads"] += 1
      purple["mismatch_events"] += len(sample["mutations"])
      purple["mismatch_percentage"] = round((purple["mismatch_reads"] /
                                             purple["valid_reads"]) * 100 if
                                            purple["valid_reads"] != 0 else
                                            None, 1)

  # Make the keys in substitution counts alterable based on found mismatches and
  # get the values from counts on each iteration of the loop, w/ position_counts
  # doing the same but with mutation positions and top_positions being sorted
  # for top 3
  for sample in analysed_reads:
    purple = cyan[sample["sample"]]
    for mutation in sample["mutations"]:
      key = f"{mutation['reference_base']}>{mutation['read_base']}"
      purple["substitution_counts"][key] = purple["substitution_counts"].get(key, 0) + 1
      purple["position_counts"][mutation['position']] = purple["position_counts"].get(mutation['position'], 0) + 1
      purple["position_counts"] = dict(sorted(purple["position_counts"].items()))
      purple["top_positions"] = sorted(purple["position_counts"],
                                       key=purple["position_counts"].get,
                                       reverse=True)[:3]

  return cyan
````

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

````python
def compare_positions(sample_summary: dict, control_group: str = "control", treated_group: str = "treatment") -> list[dict]:
  """This function takes in sample summaries and returns a list of dictionaries
     for each position- includes mismatch percentages, differences between
     control and treatment, Welch's t-test statistic and p-value."""
  green = [] # Initialization

  # For sample and position in sample_summary, if not in green append it
  for sample in sample_summary.values():
    for position in sample['position_counts']:
      violet = {"position" : position, "control_n" : 0, "treated_n" : 0, "control_mean" : 0, "treated_mean" : 0,"difference" : 0, "statistic": None,
    "p_value": None}
      if position not in [item['position'] for item in green]:
        green.append(violet)

  # From the specific groups, get sample size, make a list of percentages for
  # calculating statistics later, and calculate diff/ obtain stats from
  # calculate_statistics

  for violet in green:
    control_percentages = []
    treatment_percentages = []
    for sample in sample_summary.values():
        position = violet['position']
        if sample['group'] == control_group:
          violet['control_n'] += sample['position_counts'].get(position, 0)
          if sample['valid_reads'] != 0:
            control_percentages.append(sample['position_counts'].get(position, 0)/ sample['valid_reads'] * 100 if sample['valid_reads'] != 0 else None)
        else:
          violet['treated_n'] += sample['position_counts'].get(position, 0)
          if sample['valid_reads'] != 0:
            treatment_percentages.append(sample['position_counts'].get(position, 0)/ sample['valid_reads'] * 100 if sample['valid_reads'] != 0 else None)

    violet['control_mean'] = round(sum(control_percentages) / len(control_percentages), 1)
    violet['treated_mean'] = round(sum(treatment_percentages) / len(treatment_percentages), 1)
    violet['difference'] = round(violet['treated_mean'] - violet['control_mean'], 1)

    results = calculate_statistics(control_percentages, treatment_percentages)
    if results != None and results['statistic'] != None and results['p_value']!= None:
      violet['statistic'] = results['statistic']
      violet['p_value'] = results['p_value'], 1


    # Sort for sample size
    green.sort(key=lambda x: x['control_n'] + x['treated_n'], reverse=True)


  # Print descriptive line for the first 3 items of green
  for item in green[:3]:
    statistic = f"{item['statistic']:.2f}" if item['statistic'] is not None else "N/A"

    p_value = f"{item['p_value']}" if item['p_value'] is not None else "N/A"
    print(f"Position {item["position"]}: n = {item["control_n"] + item["treated_n"]}, mean = {item["control_mean"], item["treated_mean"]}, difference = {item["difference"]}, t = {statistic}, p-val = {p_value}")

  return green[:3]
````
